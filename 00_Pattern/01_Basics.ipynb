### Pattern 1:

```python
def pattern1(n):
    for i in range(0, n):  
        for j in range(0, n):  
            print("*", end=" ")  
        print() 
        
pattern1(5)
```

### Pattern 2:


```python
def pattern2(n):
    for i in range(0, n):  
        for j in range(0, i + 1):  
            print("*", end=" ")  
        print() 

pattern2(5)
```

### Pattern 3:

```python

def pattern3(n):
    for i in range(1, n + 1):  
        for j in range(1, i + 1):  
            print(j, end=" ")  
        print() 

pattern3(5)
```

### Pattern 4:

```python
def pattern4(n):
    for i in range(1, n + 1):  
        for j in range(1, i + 1):  
            print(i, end=" ")  
        print() 

pattern4(5)
```

### Pattern 5:

```python
def patter5(n):
    for i in range(1, n + 1):
        for j in range(1, n-i+2): 
            print("*", end=" ")
        print()
        

patter5(5)
```

### Pattern 6:

```python
def pattern6(n):
    for i in range(1, n + 1):  
        for j in range(1, n-i+2):  
            print(j, end=" ")  
        print() 

pattern6(5)
```
