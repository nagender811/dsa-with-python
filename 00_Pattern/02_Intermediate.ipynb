### Pattern 7:

```python
def pattern7(n): 
    for i in range(0, n):
        # space
        for j in range(0, n-i-1):
            print(" ", end="")
            
        # stars
        for j in range(0, i*2+1):
            print("*", end="")
            
        # space
        for j in range(0, n-i-1):
            print(" ", end="")
        print()
            
pattern7(5)
```

### Pattern 8:

```python
def pattern8(n):
    for i in range(0, n):
        # space
        for j in range(0, i):
            print(" ", end="")
        
        # stars
        for j in range(0, 2*(n-i)-1):
            print("*", end="")
        
        # space
        for j in range(0, i):
            print(" ", end="")
            
        print()

pattern8(3)                     
```
